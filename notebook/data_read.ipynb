# 춘천 ASOS 일별 기상자료 읽기

`scripts/prepare_weather.py`로 이미 전처리한 `../data/processed/weather.csv`(UTF-8, 영문 열 이름)를 읽는다.
경로는 이 노트북 위치(`notebook/`) 기준 상대경로다.

변수 설명: [docs/weather.md](../docs/weather.md)

```python
import pandas as pd

df = pd.read_csv('../data/processed/weather.csv', parse_dates=['date'])
df.info()
```

```python
df.head()
```

```python
# 결측 현황 — 빈칸은 0으로 채우지 않았다
na = df.isna().sum()
na[na > 0].to_frame('결측 일수')
```

```python
df[['temp_avg_c', 'temp_min_c', 'temp_max_c', 'precip_mm', 'humidity_avg_pct', 'wind_avg_ms']].describe().T
```

```python
# 월별 요약 (precip_mm 빈칸은 합계에서 제외, 0.0은 기록된 미량 강수)
df.groupby(df.date.dt.to_period('M')).agg(
    temp_avg=('temp_avg_c', 'mean'), temp_max=('temp_max_c', 'max'), temp_min=('temp_min_c', 'min'),
    precip_sum=('precip_mm', 'sum'), precip_recorded_days=('precip_mm', 'count'),
).round(1)
```
